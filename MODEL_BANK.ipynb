```python
# model 1.
class EMG_IMU_BiLSTM(nn.Module):
    def __init__(self, num_classes, hidden_dim=128, num_layers=1, gap_dropout=0.5):
        super().__init__()

        '''======= EMG embedding ======='''

        # 입력을 hidden_dim 크기로 projection
        self.EMG_conv1 = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=(49,1), padding=(24,0))
        self.EMG_avgpool = nn.AvgPool2d(kernel_size=(32,1))

        # Conv1D branch
        self.EMG_conv2 = nn.Conv1d(in_channels=6*16, out_channels=64, kernel_size=3, padding=1)
        self.EMG_conv3 = nn.Conv1d(in_channels=64, out_channels=64, kernel_size=3, padding=1)

        # conso branch
        self.EMG_conso_conv1 = nn.Conv1d(in_channels=16, out_channels=32, kernel_size=3, padding=1)

        # resampling timepoint to 25
        self.EMG_resample_layer = nn.Linear(16, 25)
        

        '''======= IMU embedding ======='''

        # Conv1d는 (B, C_in, L) 요구 → 입력 (B, 200, 15)를 (B, 15, 200)로 바꿔서 넣을 거야.
        self.IMU_conv1 = nn.Sequential(
            nn.Conv1d(in_channels=15, out_channels=64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.AvgPool1d(2)  # 200 -> 100
        )
        self.IMU_conv2 = nn.Sequential(
            nn.Conv1d(64, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.AvgPool1d(2)  # 100 -> 50
        )
        self.IMU_conv3 = nn.Sequential(
            nn.Conv1d(64, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.AvgPool1d(2)  # 50 -> 25
        )


        '''======= BiLSTM and classifier ======='''

        # LSTM은 (B, seq, feat) → conv 출력 (B, 64, 25)를 (B, 25, 64)로 바꿔서 넣음
        self.lstm = nn.LSTM(input_size=160, hidden_size=hidden_dim, num_layers=num_layers,
                            batch_first=True, bidirectional=True, dropout=0.0)
        self.gap = nn.AdaptiveAvgPool1d(1)  # (B, 256, L) -> (B, 256, 1)
        self.dropout = nn.Dropout(p=gap_dropout)
        self.fc = nn.Linear(hidden_dim*2, num_classes)
    
    
    def forward(self, x_EMG, x_IMU, return_feat: bool=False):
        """
        x_EMG: (batch_size, input_dim, seq_len)  (B, 6, 2048) -> (B,96,25)
        """
        x_EMG = x_EMG.permute(0, 2, 1)     # (B,2048,6)
        x_EMG = x_EMG.reshape(x_EMG.size(0), 1, x_EMG.size(1), x_EMG.size(2))  # (B,1,2048,6)

        x_EMG = F.relu(self.EMG_conv1(x_EMG))    # (B,16,2048,6)
        x_EMG = self.EMG_avgpool(x_EMG)        # (B,16,64,6)
        # reshape branch
        avg_perm = x_EMG.permute(0, 2, 3, 1)   # (B,64,6,16)
        B, T, E, C = avg_perm.shape     # B H W C
        reshape = avg_perm.reshape(B, T, E*C)   # (B,64,6*16)
        reshape = reshape.permute(0, 2, 1)      # (B,96,64)

        conv2 = F.relu(self.EMG_conv2(reshape))  # (B,64,64)
        maxpool1 = F.max_pool1d(conv2, 2)    # (B,64,32)
        conv3 = F.relu(self.EMG_conv3(maxpool1)) # (B,64,32)
        maxpool2 = F.max_pool1d(conv3, 2)    # (B,64,16)

        # conso branch
        sub = avg_perm[:,:,4,:] - avg_perm[:,:,5,:]   # (B,64,16)
        sub = sub.permute(0,2,1)   # (B,16,64)
        conso_conv1 = F.relu(self.EMG_conso_conv1(sub))   # (B,32,64)
        conso_maxpool1 = F.max_pool1d(conso_conv1, 4) # (B,32,16)

        # concat and resample to 25 timepoints
        concat_EMG = torch.cat([maxpool2, conso_maxpool1], dim=1)  # (B,96,16)
        x_EMG = self.EMG_resample_layer(concat_EMG)     # (B,96,25)


        '''
        x_IMU: (B, 200, 15) -> (B, 64, 25)
        '''
        x_IMU = x_IMU.transpose(1, 2)           # (B, 15, 200)
        x_IMU = self.IMU_conv1(x_IMU)               # (B, 64, 100)
        x_IMU = self.IMU_conv2(x_IMU)               # (B, 64, 50)
        x_IMU = self.IMU_conv3(x_IMU)               # (B, 64, 25)

        '''======= Concat ======='''
        x = torch.cat([x_EMG, x_IMU], dim=1)  # (B, 160, 25)


        x_lstm = x.transpose(1, 2)      # (B, 25, 160)
        x_lstm, _ = self.lstm(x_lstm)   # (B, 25, 256)
        x_lstm = x_lstm.transpose(1, 2) # (B, 256, 25)
        x = self.gap(x_lstm).squeeze(-1)  # (B, 256)
        x = self.dropout(x)
        logits = self.fc(x)
        return (logits, x) if return_feat else logits
```

```python
# model 62.
class EMG_IMU_Transformer(nn.Module):
    def __init__(self, num_classes, num_layers=2, hidden_dim=128, nhead=4, max_len=100):
        super().__init__()

        self.hidden_dim = hidden_dim

        '''======= EMG embedding ======='''

        # 입력을 hidden_dim 크기로 projection
        self.EMG_conv1 = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=(49,1), padding=(0,0))
        self.EMG_avgpool = nn.AvgPool2d(kernel_size=(20,1))

        # Conv1D branch
        self.EMG_conv2 = nn.Conv1d(in_channels=6*16, out_channels=64, kernel_size=3, padding=1)
        self.EMG_conv3 = nn.Conv1d(in_channels=64, out_channels=64, kernel_size=3, padding=1)

        # conso branch
        self.EMG_conso_conv1 = nn.Conv1d(in_channels=16, out_channels=64, kernel_size=3, padding=1)
        

        '''======= IMU embedding ======='''

        # Conv1d는 (B, C_in, L) 요구 → 입력 (B, 200, 15)를 (B, 15, 200)로 바꿔서 넣을 거야.
        self.IMU_conv1 = nn.Sequential(
            nn.Conv1d(in_channels=15, out_channels=64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.AvgPool1d(2)  # 200 -> 100
        )
        self.IMU_conv2 = nn.Sequential(
            nn.Conv1d(64, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.AvgPool1d(2)  # 100 -> 50
        )
        self.IMU_conv3 = nn.Sequential(
            nn.Conv1d(64, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.AvgPool1d(2)  # 50 -> 25
        )


        # 위치 임베딩
        self.pos_embedding = nn.Embedding(max_len, hidden_dim)

        # Corss Modality Transformer Encoder
        encoder_layer = PreNormTransformerEncoderLayer(d_model=hidden_dim, nhead=nhead, dim_feedforward=hidden_dim*4)
        self.cross_transformer = CrossModalityTransformerEncoder(encoder_layer, num_layers=num_layers)

        # Self Attention transformer Encoder
        self.EMG_transformer = PreNormTransformerEncoder(encoder_layer, num_layers=num_layers)
        self.IMU_transformer = PreNormTransformerEncoder(encoder_layer, num_layers=num_layers)

        # 분류기
        self.classifier = nn.Linear(hidden_dim, num_classes)


    def forward(self, x_EMG, x_IMU, return_feat: bool=False):
        """
        x_EMG: (batch_size, input_dim, seq_len)  (B, 6, 2048) -> (B,96,25)
        """
        x_EMG = x_EMG.permute(0, 2, 1)     # (B,2048,6)
        x_EMG = x_EMG.reshape(x_EMG.size(0), 1, x_EMG.size(1), x_EMG.size(2))  # (B,1,2048,6)

        x_EMG = F.relu(self.EMG_conv1(x_EMG))    # (B,16,2000,6)
        x_EMG = self.EMG_avgpool(x_EMG)        # (B,16,100,6)
        # reshape branch
        avg_perm = x_EMG.permute(0, 2, 3, 1)   # (B,100,6,16)
        B, T, E, C = avg_perm.shape     # B H W C
        reshape = avg_perm.reshape(B, T, E*C)   # (B,100,6*16)
        reshape = reshape.permute(0, 2, 1)      # (B,96,100)

        conv2 = F.relu(self.EMG_conv2(reshape))  # (B,32,100)
        maxpool1 = F.max_pool1d(conv2, 2)    # (B,32,50)
        conv3 = F.relu(self.EMG_conv3(maxpool1)) # (B,64,50)
        maxpool2 = F.max_pool1d(conv3, 2)    # (B,64,25)

        # conso branch
        sub = avg_perm[:,:,4,:] - avg_perm[:,:,5,:]   # (B,100,16)
        sub = sub.permute(0,2,1)   # (B,16,100)
        conso_conv1 = F.relu(self.EMG_conso_conv1(sub))   # (B,64,100)
        conso_maxpool1 = F.max_pool1d(conso_conv1, 4) # (B,64,25)

        # concat and resample to 25 timepoints
        x_EMG = maxpool2 + conso_maxpool1  # (B,64,25)


        '''
        x_IMU: (B, 200, 15) -> (B, 64, 25)
        '''
        x_IMU = x_IMU.transpose(1, 2)           # (B, 15, 200)
        x_IMU = self.IMU_conv1(x_IMU)               # (B, 64, 100)
        x_IMU = self.IMU_conv2(x_IMU)               # (B, 64, 50)
        x_IMU = self.IMU_conv3(x_IMU)               # (B, 64, 25)

        '''======= Concat ======='''

        x_EMG = x_EMG.permute(0, 2, 1)     # (B,25,64)
        x_IMU = x_IMU.permute(0, 2, 1)     # (B,25,64)

        batch_size, seq_len, _ = x_IMU.size()

        # 위치 임베딩 추가
        positions = torch.arange(seq_len, device=x_IMU.device).unsqueeze(0).expand(batch_size, seq_len)
        x_EMG = x_EMG + self.pos_embedding(positions)
        x_IMU = x_IMU + self.pos_embedding(positions)

        # Transformer 입력은 (seq_len, batch, hidden_dim)
        x_EMG = x_EMG.transpose(0, 1)
        x_IMU = x_IMU.transpose(0, 1)

        # Encoder 통과
        cross_EMG, cross_IMU = self.cross_transformer(src1=x_EMG,src2=x_IMU)   # (seq_len, batch, hidden_dim)
        self_EMG = self.EMG_transformer(x_EMG)
        self_IMU = self.IMU_transformer(x_IMU)

        # concatenate outputs
        out = cross_EMG + cross_IMU + self_EMG + self_IMU    # (seq_len, batch, hidden_dim)

        # [CLS] 대용: 첫 번째 토큰 사용
        cls_out = out.mean(dim=0)   # (batch, hidden_dim)

        # 분류기
        return self.classifier(cls_out)
```
